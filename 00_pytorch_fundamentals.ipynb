### Random Tensors

WHy random tensors?

Random tensors are important because the way many neural networks learn is that the strat with tensors full of random numbers and then adjust those random numbers to better represent the data.

`Start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers`

random tensors: https://docs.pytorch.org/docs/stable/generated/torch.rand.html

### Tensor datatypes
Understanding tensor datatypes is very important because it is one of the 3 big errors that is often encountered with pytorch and deep learning.

1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not right device

Data types docummentation: https://docs.pytorch.org/docs/stable/tensor_attributes.html#torch.dtype


```python
# Float 32 tensor.
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                                dtype=None, # tensors are of type torch.float32 by default
                                device=None,  # what device is the tensor on? cpu, gpu, tpu, cuda, etc.
                                requires_grad=False # track gradients with this tensor
                                )
float_32_tensor
```

```python
float_32_tensor
```

```python
# convert float_32_tensor to a tensor of type torch.float16
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_16_tensor * float_32_tensor
```

```python
complex_tensor = torch.tensor([3-1j, 6+2j, 9j],
                                dtype=None,
                                device=None,
                                requires_grad=False
                                )
complex_tensor
```

```python
complex_tensor.dtype
```

```python
complex_tensor * float_16_tensor
```

```python
# 2.06.57
```

### Reshaping, stacking, squezzing and unsqueezing tensors

* Reshaping --> reshapes an input tensor to a defined shape
* View --> return a view of an input tensor of certain shape but keep the same memory as original tensor
* Stacking --> combine multiple tensors on top of ecah other (vstack) or side by side (hstack)
* Squeeze --> removes all 1 dimension from a tensor
* Unsqueeze --> add a 1 dimension to a target tensor
* Permute - return a view of the input with dimensions permuted (swapped) in a certain way

```python
# Let's start over
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Reshape
# x_reshaped = x.reshape(1, 7) --> would fail because the only dimension we have has 9 elements
# for reshape to succeed the elements available must fit into the new shape
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
#x_reshaped = x.reshape(9, 1)
#x_reshaped, x_reshaped.shape
```

```python
another_x = torch.arange(1., 11.)
another_x, another_x.shape
```

```python
another_x_reshaped = another_x.reshape(5, 2)
another_x_reshaped, another_x_reshaped.shape
```

```python
another_x_reshaped = another_x.reshape(2, 5)
another_x_reshaped, another_x_reshaped.shape
```

```python
# View
z = x.view(1,9)    # --> z will use the same memory as x. any operation on the view of tensor will also affect the tensor
z, z.shape
```

```python
z[:, 0] = 5
z, x
```

```python
# Stacking
x_stacked = torch.stack([x,x,x,x], dim=1)
x_stacked
```

```python
_stacked = torch.vstack([x,x,x,x])   # vertical stacking
_stacked
```

```python
_stacked = torch.hstack([x,x,x,x])   # horizontal
_stacked
```

```python
# Squize -->  Removes dimension 0f size 1 from a given tensor
x_reshaped
```

```python
x_reshaped.shape
```

```python
x_squeezed = x_reshaped.squeeze()
x_squeezed
```

```python
x_squeezed.shape
```

```python
# Unsqueeze --> adds a single dimension to a tensor at the specified dimension
print(f"previous targer: {x_squeezed}")
print(f"previous shape: {x_squeezed.shape}")

x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# Permute   --> rearranges elements in a tensor to follow a given order of dimensions
x_original = torch.rand(224, 224, 3)     # sample image tensor [height, width, colour_channels]

x_permuted = x_original.permute(2,0,1)
print(f"original shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")

```

```python
# Note that the permute version of a tensor shares the same memory with the original tensor
print(x_original)
print(x_permuted)
x_original[0,0,0] = 1988
print(x_original)
print(x_permuted)
```
