### Random Tensors

WHy random tensors?

Random tensors are important because the way many neural networks learn is that the strat with tensors full of random numbers and then adjust those random numbers to better represent the data.

`Start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers`

random tensors: https://docs.pytorch.org/docs/stable/generated/torch.rand.html

```python
# Create random tensor of  size (3, 4)
random_tensor = torch.rand(3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
ANOTHERTENSOR = torch.rand(10,10,10)
ANOTHERTENSOR
```

```python
ANOTHERTENSOR.ndim, ANOTHERTENSOR.shape
```

```python
# Create a random tensor with similar shape to an image tensor
random_image_tensor = torch.rand(size=(224, 224, 3)) # height, width, colour channel
random_image_tensor.shape, random_image_tensor.ndim
```

### Tensor datatypes
Understanding tensor datatypes is very important because it is one of the 3 big errors that is often encountered with pytorch and deep learning.

1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not right device

Data types docummentation: https://docs.pytorch.org/docs/stable/tensor_attributes.html#torch.dtype


```python
# Float 32 tensor.
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                                dtype=None, # tensors are of type torch.float32 by default
                                device=None,  # what device is the tensor on? cpu, gpu, tpu, cuda, etc.
                                requires_grad=False # track gradients with this tensor
                                )
float_32_tensor
```

```python
float_32_tensor
```

```python
# convert float_32_tensor to a tensor of type torch.float16
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_16_tensor * float_32_tensor
```

```python
complex_tensor = torch.tensor([3-1j, 6+2j, 9j],
                                dtype=None,
                                device=None,
                                requires_grad=False
                                )
complex_tensor
```

```python
complex_tensor.dtype
```
